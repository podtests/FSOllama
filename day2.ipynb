### How to create tools in Ollama

```python
import os
from ollama import Client, Message

def get_weather_city(city: str) -> str:
    """Get the current weather for a city
    
    Args:
        city: The name of the city for which to get weather information
    
    Returns:
        A string describing the weather in the specified city
    """
    return f"The weather in {city} is sunny."


def calculate(expression: str) -> str:
    """Calculate the result of a mathematical expression
    
    Args:
        expression: A string representing a mathematical expression
    
    Returns:
        A string representing the result of the calculation
    """
    return str(eval(expression))




def callPodTestAgent(promtinput: str) -> str:
    msg = []
    msgnew = Message(role="user", content=promtinput)   
    msg.append(msgnew) #1

    Tools= {
        'get_weather_city': get_weather_city,
        'calculate': calculate
    }

    client = Client(host="https://ollama.com",              
                    headers={'Authorization': 'Bearer ' + 'f3f5411d40da47bc96283a51b1a76985.GDkd-mtPdmc9ljAsOaIOBgPb'}
                    )
    while True:
        res = client.chat(
                    model='qwen3-coder-next:cloud',
                    messages=msg,
                    tools=[get_weather_city, calculate]
                    #think=False            
                    )
        msg.append(res.message) #2

        if not res.message.tool_calls:
            return res.message.content
        
        #ToolCall(function=Function(name='get_weather_city', arguments={'city': 'Delhi', }))
        for tool_call in res.message.tool_calls:
            fn_name = tool_call.function.name
            fn_args = tool_call.function.arguments   # {'city': 'Delhi', 'time': 'now'}       
            result = Tools[fn_name](**fn_args)
            msg.append({'role': 'tool','name': fn_name,'content': result})  #3, #4
        
        #print(msg) 

res = callPodTestAgent("What is the weather in Delhi right now? Also tell me the capital of Australia & also tell me the calulated value of  6*7")
print(res)

#- The weather in Delhi is currently sunny.  
#- The capital of Australia is Canberra.  
#- The calculated value of $6 \times 7$ is 42.



```

### langchain code for creating loop ourself.

```python
import os
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage, ToolMessage
from langchain_ollama import ChatOllama

# ── Tools ──────────────────────────────────────────────────────────────────────

@tool
def get_weather_city(city: str) -> str:
    """Get the current weather for a city.

    Args:
        city: The name of the city for which to get weather information.
    """
    return f"The weather in {city} is sunny."


@tool
def calculate(expression: str) -> str:
    """Calculate the result of a mathematical expression.

    Args:
        expression: A string representing a mathematical expression.
    """
    return str(eval(expression))


# ── Agent ──────────────────────────────────────────────────────────────────────

def call_pod_test_agent(prompt_input: str) -> str:
    tools = [get_weather_city, calculate]
    tools_by_name = {t.name: t for t in tools}

    llm = ChatOllama(
        model="qwen3-coder-next:cloud",
        base_url="https://ollama.com",
        client_kwargs={
            "headers": {
                "Authorization": "Bearer f3f5411d40da47bc96283a51b1a76985.GDkd-mtPdmc9ljAsOaIOBgPb"
            }
        },
    ).bind_tools(tools)  # bind tools to the model once

    messages = [HumanMessage(content=prompt_input)]  # #1

    while True:
        response = llm.invoke(messages)               # #2
        messages.append(response)

        if not response.tool_calls:
            return response.content

        # Execute each tool call and append results
        for tool_call in response.tool_calls:         # #3
            fn_name = tool_call["name"]
            fn_args = tool_call["args"]

            result = tools_by_name[fn_name].invoke(fn_args)

            messages.append(                          # #4
                ToolMessage(
                    content=str(result),
                    tool_call_id=tool_call["id"],     # required by LangChain
                    name=fn_name,
                )
            )


# ── Run ────────────────────────────────────────────────────────────────────────

res = call_pod_test_agent(
    "What is the weather in Delhi right now? "
    "Also tell me the capital of Australia & "
    "also tell me the calculated value of 6*7"
)
print(res)
```
